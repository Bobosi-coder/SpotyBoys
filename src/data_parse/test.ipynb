```python
import os
from pathlib import Path
import sys  
```

```python
def read_file_8_rows(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        data = []
        for i in range(8):
            data.append(repr(f.readline()))
    return data
```

```python
data_path = Path("../../data/30music")
output_file = Path("30_music_data_summary.txt")

with open(output_file, "w", encoding="utf-8")as out_f:
    sub_dirs = [dir for dir in data_path.iterdir() if dir.is_dir() and not dir.name.startswith(".")]

    for sub_dir in sub_dirs:
        header = f"\n>>> This is {sub_dir.name} directory:\n"
        print(header, end="")
        out_f.write(header)

        files = [f for f in sub_dir.iterdir() if f.is_file() and not f.name.startswith(".")]
        
        for i, file_path in enumerate(files):
            file_info = f"  [{i + 1}] File: {file_path.name}\n"
            print(file_info, end="")
            out_f.write(file_info)

            rows = read_file_8_rows(file_path)
            for row in rows:
                out_f.write(f"      {row}\n")
                print(f"      {row}")

            divider = "  " + "="*30 + "\n"
            print(divider, end="")
            out_f.write(divider)
```

```python
import pandas as pd
```

```python
# 文件路径
file_path = '/Users/breeze./Desktop/Project/MLOps/data/processed/session_tracks_filtered.csv'

# 初始化计数器
total_rows = 0
label_counts = {'positive': 0, 'skip': 0, 'neutral': 0}

# 设置分块大小（例如每 100,000 行处理一次，根据内存情况可调整）
chunk_size = 100000

try:
    # 使用 chunksize 分块读取，只读取 label 列以节省内存
    for chunk in pd.read_csv(file_path, usecols=['label'], chunksize=chunk_size):
        # 统计当前块的总行数
        total_rows += len(chunk)
        
        # 统计 label 列中各值的出现次数
        counts = chunk['label'].value_counts()
        
        # 累加到全局计数器
        for label in label_counts.keys():
            if label in counts:
                label_counts[label] += counts[label]

    # 输出结果
    print(f"统计完成！")
    print(f"总数据行数: {total_rows}")
    print(f"Positive 数量: {label_counts['positive']}")
    print(f"Negative 数量: {label_counts['skip']}")
    print(f"Neutral 数量: {label_counts['neutral']}")
    print(label_counts['positive'] + label_counts['skip'] + label_counts['neutral'])

except FileNotFoundError:
    print(f"错误：找不到文件，请检查路径是否正确：{file_path}")
except Exception as e:
    print(f"发生错误：{e}")
```
